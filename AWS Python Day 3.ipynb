```python
students = {}


def show_menu():
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Exit")


def add_students(student_id, name, marks):
    students[student_id] = {"Name": name, "Marks": marks}
    print(name, "added successfully")


def view_students():
    if len(students) == 0:
        print("No students found")
    else:
        for sid, details in students.items():
            print("ID:", sid, "| Name:", details["Name"], "| Marks:", details["Marks"])
            # print(sid, details)


def search_students(student_id):
    details = students.get(student_id)
    if details:
        print("Name:", details["Name"], "| Marks:", details["Marks"])
    else:
        print("Student not found")


def update_marks(student_id, new_marks):
    if student_id in students:
        students[student_id]["Marks"] = new_marks
        print("Marks updated successfully for", students[student_id]["Name"])
    else:
        print("Student not found")

def run_program():
    while True:
        show_menu()
        print("5. Update Student Marks")  # New option
        try:
            choice = int(input("Enter your choice: "))
        except ValueError:
            print("Invalid input! Please enter a number (1–5).")
            continue

        if choice == 1:
            student_id = input("Enter your student_id: ")
            name = input("Enter your name: ")

            while True:
                try:
                    marks = int(input("Enter your marks (0–100): "))
                    if 0 <= marks <= 100:
                        break
                    else:
                        print("Marks must be between 0 and 100. Please try again.")
                except ValueError:
                    print("Invalid input! Please enter a number between 0 and 100.")

            add_students(student_id, name, marks)
            print("Student added successfully")

        elif choice == 2:
            view_students()

        elif choice == 3:
            student_id = input("Enter your student_id: ")
            search_students(student_id)

        elif choice == 4:
            confirm = input("Are you sure you want to exit? (y/n): ").lower()
            if confirm == "y":
                print("Exiting program...")
                break
            else:
                print("Returning to menu...")

        elif choice == 5:
            student_id = input("Enter the student_id to update marks: ")
            while True:
                try:
                    new_marks = int(input("Enter new marks (0–100): "))
                    if 0 <= new_marks <= 100:
                        break
                    else:
                        print("Marks must be between 0 and 100. Please try again.")
                except ValueError:
                    print("Invalid input! Please enter a number between 0 and 100.")
            update_marks(student_id, new_marks)

        else:
            print("Enter valid option:")
run_program()
```
