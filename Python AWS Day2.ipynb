```python
first_number = int(input("Enter first number "))
second_number = int(input("Enter second number "))
operator_chosen = input("Operator_chosen ")
if operator_chosen == "+" :
    print(first_number + second_number)
elif operator_chosen == "-":
    print(first_number - second_number)
elif operator_chosen == "*":
    print(first_number * second_number)
elif operator_chosen == "/":
    print(first_number / second_number)
else :
    print("choose =,-,*,or /")
    
```

```python
##ATM Code
Card_Validity = True
Correct_Pin = 1234
Account_Balance = 10000

if(Card_Validity) == True :
    Entered_Pin = int(input("Enter your PIN"))  
    if(Entered_Pin == Correct_Pin):
        Withdrawl_Amount = int(input("Enter your withdrawl amount"))        
        if(Withdrawl_Amount) < Account_Balance:
            print("Collect your money please")
        else:
            print("Please enter amount less than account balance")
    else :
            print("Enter Correct Pin")   
else: 
    print("Enter Valid Card")
```
